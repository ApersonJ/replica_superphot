```python
import time
import pandas as pd
from alerce.core import Alerce

client = Alerce()
muestra = pd.read_csv("../data/muestra.csv")
prueba = muestra.sample(20, random_state=0)

tiempos = []
for oid in prueba["NAME"]:
    t0 = time.time()
    try:
        det = client.query_detections(oid, survey="ztf", format="pandas")
        resultado = len(det)
    except Exception as e:
        resultado = f"error: {e}"
    tiempos.append(time.time() - t0)
    print(oid, resultado, round(tiempos[-1], 2), "s")

print("promedio:", round(sum(tiempos) / len(tiempos), 2), "s")
```
