SN Ia-91T no se muestra en el paper, se incluira en la categoria SN Ia.

```python
SEED = 42
df = pd.read_csv("../data/training_set.csv")

MAPA = {
    "SN Ia": "SN Ia", "SN Ia-91T-like": "SN Ia", "SN Ia-91T": "SN Ia",
    "SN Ia-91bg-like": "SN Ia", "SN Ia-CSM": "SN Ia",
    "SN II": "SN II", "SN IIP": "SN II", "SN IIL": "SN II",
    "SN Ib": "SN Ib/c", "SN Ic": "SN Ib/c", "SN Ic-BL": "SN Ib/c", "SN Ib/c": "SN Ib/c",
    "SN IIn": "SN IIn", "SLSN-II": "SN IIn",
    "SLSN-I": "SLSN-I",
}
df["CLASE"] = df["LABEL"].map(MAPA)
print("Fuera de las 5 clases:", df["CLASE"].isna().sum())
df = df.dropna(subset=["CLASE"])
print(df["CLASE"].value_counts())

muestra = df.reset_index(drop=True)   
print(muestra["CLASE"].value_counts())
print(len(muestra), "objetos")
muestra.to_csv("../data/muestra.csv", index=False)
```
